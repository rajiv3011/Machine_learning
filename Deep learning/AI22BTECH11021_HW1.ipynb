```python
x=[1,2,3,4]
px1=[1/2,1/4,1/8,1/8]

px2=[1/2,1/4,0,1/4]


def entropy(px):
  Entropy=0
  for i in range(len(px)):
    if(px[i]==0):
      continue
    Entropy=Entropy-px[i]*np.log2(px[i])

  return Entropy

e1=entropy(px1)
e2=entropy(px2)

print("Entropy without zero probability",e1)
print("Entropy with zero probability",e2)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import cv2

image_path = "/content/drive/MyDrive/misc/5.1.09.tiff"
img = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

def normalized_histogram(img):
  hist=np.zeros(256)

  for i in range(len(img)):
    for j in range(len(img[0])):
      hist[img[i][j]]+=1

  hist=hist/np.sum(hist)

  return hist

hist=normalized_histogram(img)


plt.bar(range(len(hist)),hist)
plt.title('Normalized Histogram')
plt.xlabel('Pixel Value')
plt.ylabel('Frequency')
plt.show()
```

```python
image_path = "/content/drive/MyDrive/left.png"
left = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

image_path = "/content/drive/MyDrive/right.png"
right = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

def Joint_histogram(img1,img2):
  Joint_histogram=np.zeros((256,256))

  for i in range(len(img1)):
    for j in range(len(img1[0])):
      Joint_histogram[img1[i][j]][img2[i][j]]+=1

  Joint_histogram=Joint_histogram/np.sum(Joint_histogram)

  return Joint_histogram

Joint_histogram=Joint_histogram(left,right)

plt.imshow(Joint_histogram, cmap='gray')
plt.colorbar()
plt.title('Joint Histogram')
plt.xlabel('Intensity - Image 2')
plt.ylabel('Intensity - Image 1')
plt.show()
```

(b)

```python
def joint_entropy(Joint_histogram):
  Joint_entropy=0

  for i in range(len(Joint_histogram)):
    for j in range(len(Joint_histogram[0])):
      if(Joint_histogram[i][j]==0):
        continue
      Joint_entropy=Joint_entropy-Joint_histogram[i][j]*np.log2(Joint_histogram[i][j])

  return Joint_entropy


print("Joint entropy is :",joint_entropy(Joint_histogram))
```

```python
def compute_conditional_pmf(joint_pmf, conditioning_index, conditioning_value):
  if conditioning_index == 0:
    # P(X | Y = y)
    marginal_pmf = np.sum(joint_pmf, axis=0)
    conditional_pmf = joint_pmf[:, conditioning_value] / marginal_pmf[conditioning_value]
  else:
    # P(Y | X = x)
    marginal_pmf = np.sum(joint_pmf, axis=1)
    conditional_pmf = joint_pmf[conditioning_value, :] / marginal_pmf[conditioning_value]

  return conditional_pmf

x=compute_conditional_pmf(Joint_histogram,0,1)
```
